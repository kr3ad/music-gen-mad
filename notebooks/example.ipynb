```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

print('✅ All imports successful!')
print(f'NumPy: {np.__version__}')
print(f'Pandas: {pd.__version__}')
```

```python
# Create and explore sample data
X, y = make_classification(n_samples=100, n_features=4, random_state=42)
df = pd.DataFrame(X, columns=[f'feature_{i+1}' for i in range(4)])
df['target'] = y

print('Data shape:', df.shape)
print('\nFirst few rows:')
print(df.head())
```

```python
# Visualize
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df.plot(x='feature_1', y='feature_2', kind='scatter', ax=axes[0])
axes[0].set_title('Feature 1 vs Feature 2')

df['target'].value_counts().plot(kind='bar', ax=axes[1])
axes[1].set_title('Target Distribution')

plt.tight_layout()
plt.show()
```
